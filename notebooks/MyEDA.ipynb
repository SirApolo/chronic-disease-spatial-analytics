```python
lista_arquivos = os.listdir("../data/raw/bases_auxiliares/")
lista_arquivos = [csv for csv in lista_arquivos if csv.endswith(".csv")]
dict_csv = {}
parse_options = pcsv.ParseOptions(delimiter=";")
for csv in lista_arquivos:
    dict_csv.update({csv[:-4]: pcsv.read_csv(os.path.join("../data/raw/bases_auxiliares/", csv), parse_options=parse_options)})
```

```python
dict_csv
```

```python
for key, value in dict_csv.items():
    df_sih.join(keys=key, right_table=value, right_keys="value", right_suffix="_CSV")
```
